## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd
import requests
import io
from google.colab import userdata

# Get Hugging Face token from Colab Secrets
hf_token = userdata.get('HF_TOKEN')

# Parquet dataset URL
url = "https://huggingface.co/datasets/FlyRank/internship-warehouse/resolve/main/fact_content_daily_performance_sample.parquet"
headers = {"Authorization": f"Bearer {hf_token}"}

response = requests.get(url, headers=headers)

if response.status_code == 200:
    df = pd.read_parquet(io.BytesIO(response.content))

    print("Dataset Columns:", list(df.columns))
    print(f"Total Daily Sample Records: {len(df):,}")

    # Check for exact column names or fallbacks
    click_col = [c for c in df.columns if 'click' in c.lower()]
    imp_col = [c for c in df.columns if 'impression' in c.lower()]

    if click_col:
        print(f"Average Daily Clicks per Record ({click_col[0]}): {df[click_col[0]].mean():.2f}")
    if imp_col:
        print(f"Average Daily Impressions per Record ({imp_col[0]}): {df[imp_col[0]].mean():.2f}")
else:
    print(f"Failed to fetch dataset. HTTP Status: {response.status_code}")
```
